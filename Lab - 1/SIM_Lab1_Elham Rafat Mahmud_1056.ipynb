# **Scalar** &amp; Array Operations



```python
import numpy as np

# Sample scalar and array inputs
scalar = 3.7
arr = np.array([1.2, 2.7, 3.5, 4.1, 5.9, 6.3])  # 6 elements

# Trigonometric functions
sin_values = np.sin(arr)
cos_values = np.cos(arr)
tan_values = np.tan(arr)
asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
acos_values = np.arccos(np.clip(arr / np.max(arr), -1, 1))
atan_values = np.arctan(arr)

# Exponential and logarithm
exp_values = np.exp(arr)
log_values = np.log(arr)

# Absolute value and square root
abs_values = np.abs(arr)
sqrt_values = np.sqrt(arr)

# Remainder when divided by 2
rem_values = np.remainder(arr, 2)

# Rounding functions
round_values = np.round(arr)
floor_values = np.floor(arr)
ceil_values = np.ceil(arr)

# Display results
print("scalar:", scalar)
print("arr:", arr)
print("sin:", sin_values)
print("cos:", cos_values)
print("tan:", tan_values)
print("asin:", asin_values)
print("acos:", acos_values)
print("atan:", atan_values)
print("exp:", exp_values)
print("log:", log_values)
print("abs:", abs_values)
print("sqrt:", sqrt_values)
print("rem:", rem_values)
print("round:", round_values)
print("floor:", floor_values)
print("ceil:", ceil_values)
```

# **2×2 Matrix Calculations**

```python
import numpy as np

# Creating a 3x2 matrix with custom values
matrix = np.array([[1.5, 2.3], [3.8, -4.1], [5.0, 6.6]])

# Trigonometric functions
sin_values = np.sin(matrix)
cos_values = np.cos(matrix)
tan_values = np.tan(matrix)
asin_values = np.arcsin(np.clip(matrix / np.max(np.abs(matrix)), -1, 1))  # safe normalization
acos_values = np.arccos(np.clip(matrix / np.max(np.abs(matrix)), -1, 1))
atan_values = np.arctan(matrix)

# Exponential and logarithm
exp_values = np.exp(matrix)
log_values = np.log(np.abs(matrix))  # abs since log of negatives is undefined

# Absolute value and square root
abs_values = np.abs(matrix)
sqrt_values = np.sqrt(np.abs(matrix))  # abs since sqrt of negatives is undefined (real domain)

# Remainder when divided by 2
rem_values = np.remainder(matrix, 2)

# Rounding functions
round_values = np.round(matrix)
floor_values = np.floor(matrix)
ceil_values = np.ceil(matrix)

# Display results
print("Matrix:\n", matrix)
print("\nsin:\n", sin_values)
print("\ncos:\n", cos_values)
print("\ntan:\n", tan_values)
print("\nasin:\n", asin_values)
print("\nacos:\n", acos_values)
print("\natan:\n", atan_values)
print("\nexp:\n", exp_values)
print("\nlog (natural, abs values):\n", log_values)
print("\nabs:\n", abs_values)
print("\nsqrt (of abs values):\n", sqrt_values)
print("\nrem (remainder when divided by 2):\n", rem_values)
print("\nround:\n", round_values)
print("\nfloor:\n", floor_values)
print("\nceil:\n", ceil_values)
```

# **Vector Functions**

Core Operations (Static 2×3 Matrix)

```python
import numpy as np

mat = np.array([[3.2, 1.5, 4.8],
                 [1.1, 5.9, 2.3]])

# Max & min with indices
max_value = np.max(mat)
max_index = np.unravel_index(np.argmax(mat), mat.shape)
min_value = np.min(mat)
min_index = np.unravel_index(np.argmin(mat), mat.shape)

# Length
num_rows = len(mat)
total_elements = mat.size

# Sorting
sorted_rowwise_desc = -np.sort(-mat, axis=1)
sorted_flatten_desc = -np.sort(-mat.flatten())

# Stats
sum_values = np.sum(mat)
prod_values = np.prod(mat)
median_value = np.median(mat)
mean_value = np.mean(mat)
std_dev = np.std(mat)

# Display results
print("Matrix:\n", mat)
print("\nMax value:", max_value, "at index (row, col) =", max_index)
print("Min value:", min_value, "at index (row, col) =", min_index)
print("Number of rows (len):", num_rows, "| Total elements (size):", total_elements)
print("\nRow-wise sorted (descending):\n", sorted_rowwise_desc)
print("\nEntire matrix sorted (descending, flattened):\n", sorted_flatten_desc)
print("\nSum of elements:", sum_values)
print("Product of elements:", prod_values)
print("Median value:", median_value)
print("Mean value:", mean_value)
print("Standard deviation:", std_dev)
```

Bonus Part

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)  # reproducible results
mat_random = np.round(np.random.uniform(low=1.0, high=6.0, size=(2, 3)), 2)

max_value = np.max(mat_random)
max_index = np.unravel_index(np.argmax(mat_random), mat_random.shape)
min_value = np.min(mat_random)
min_index = np.unravel_index(np.argmin(mat_random), mat_random.shape)

sorted_rowwise_desc = -np.sort(-mat_random, axis=1)
sorted_flatten_desc = -np.sort(-mat_random.flatten())

sum_values = np.sum(mat_random)
prod_values = np.prod(mat_random)
median_value = np.median(mat_random)
mean_value = np.mean(mat_random)
std_dev = np.std(mat_random)

# Display results
print("Randomly generated matrix:\n", mat_random)
print("\nMax value:", max_value, "at index (row, col) =", max_index)
print("Min value:", min_value, "at index (row, col) =", min_index)
print("\nRow-wise sorted (descending):\n", sorted_rowwise_desc)
print("\nEntire matrix sorted (descending, flattened):\n", sorted_flatten_desc)
print("\nSum of elements:", sum_values)
print("Product of elements:", prod_values)
print("Median value:", median_value)
print("Mean value:", mean_value)
print("Standard deviation:", std_dev)

# Heatmap visualization
sns.heatmap(mat_random, annot=True, fmt=".2f", cmap="viridis", cbar=True,
            xticklabels=["col0", "col1", "col2"], yticklabels=["row0", "row1"])
plt.title("Heatmap of 2x3 Random Matrix")
plt.show()
```
