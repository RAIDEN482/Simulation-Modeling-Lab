```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("--- Part A: Uniform Distribution ---")

uniform_waiting_times = np.random.uniform(low=3, high=15, size=1000)


print("First 10 uniform waiting times:", uniform_waiting_times[:10])


uniform_mean = np.mean(uniform_waiting_times)
uniform_median = np.median(uniform_waiting_times)
uniform_std = np.std(uniform_waiting_times)

print(f"Mean (Uniform): {uniform_mean:.2f} minutes")
print(f"Median (Uniform): {uniform_median:.2f} minutes")
print(f"Standard Deviation (Uniform): {uniform_std:.2f} minutes")

print("\n--- Part B: Normal Distribution ---")

normal_waiting_times = np.random.normal(loc=9, scale=2, size=1000)


normal_waiting_times = np.maximum(0, normal_waiting_times)

# Calculate and print: Mean, Median, Standard deviation, Maximum waiting time
normal_mean = np.mean(normal_waiting_times)
normal_median = np.median(normal_waiting_times)
normal_std = np.std(normal_waiting_times)
normal_max = np.max(normal_waiting_times)

print(f"Mean (Normal): {normal_mean:.2f} minutes")
print(f"Median (Normal): {normal_median:.2f} minutes")
print(f"Standard Deviation (Normal): {normal_std:.2f} minutes")
print(f"Maximum Waiting Time (Normal): {normal_max:.2f} minutes")

print("\n--- Part C: Visualization ---")
# Create one figure containing two subplots:
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Subplot 1 -> Histogram of the uniform-distribution data
axes[0].hist(uniform_waiting_times, bins=20, edgecolor='black', alpha=0.7)
axes[0].set_title('Histogram of Uniform Waiting Times')
axes[0].set_xlabel('Waiting Time (minutes)')
axes[0].set_ylabel('Frequency')

# Subplot 2 -> Histogram of the normal-distribution data
axes[1].hist(normal_waiting_times, bins=20, edgecolor='black', alpha=0.7, color='orange')
axes[1].set_title('Histogram of Normal Waiting Times')
axes[1].set_xlabel('Waiting Time (minutes)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

print("\n--- Part D: Reproducibility ---")



np.random.seed(32)
matrix_seeded_1 = np.random.randint(low=1, high=51, size=(3, 3))
print("Matrix with seed (first run, seed=32):\n", matrix_seeded_1)

#same seed.
np.random.seed(32)
matrix_seeded_2 = np.random.randint(low=1, high=51, size=(3, 3))
print("\nMatrix with same seed (second run, seed=32):\n", matrix_seeded_2)

#without the seed.
matrix_unseeded = np.random.randint(low=1, high=51, size=(3, 3))
print("\nMatrix without setting seed:\n", matrix_unseeded)
```
